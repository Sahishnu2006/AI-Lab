```python
import pandas as pd
import re
import numpy as np


# ---------------------------------------------------------
# 1. Text normalization
# ---------------------------------------------------------

def clean_str(text):
    """Normalize text while preserving alphanumeric information."""
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    # Normalize common separators
    text = text.replace("&", " and ")

    # Remove punctuation
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Collapse whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


# ---------------------------------------------------------
# 2. Postal code extraction
# ---------------------------------------------------------

def extract_postal_code(address):
    """
    Extract common postal/ZIP code formats.

    Handles examples such as:
        10001
        10001-1234
        SW1A 1AA
        M5V 3A8
        560001
    """

    if pd.isna(address):
        return ""

    address = str(address).upper()

    patterns = [
        # US ZIP
        r"\b\d{5}(?:-\d{4})?\b",

        # UK postcode
        r"\b[A-Z]{1,2}\d[A-Z\d]?\s*\d[A-Z]{2}\b",

        # Canada
        r"\b[A-Z]\d[A-Z][ -]?\d[A-Z]\d\b",

        # Generic 4-6 digit postal code
        r"\b\d{4,6}\b",
    ]

    for pattern in patterns:
        match = re.search(pattern, address)

        if match:
            return re.sub(r"\s+", "", match.group(0)).upper()

    return ""


# ---------------------------------------------------------
# 3. Main preprocessing pipeline
# ---------------------------------------------------------

def preprocess_entities(df):
    df = df.copy()

    # Make sure required fields exist
    required_cols = [
        "business_name",
        "business_address",
        "country"
    ]

    for col in required_cols:
        if col not in df.columns:
            df[col] = ""

    # -----------------------------------------------------
    # Fill missing values
    # -----------------------------------------------------

    df["business_name"] = (
        df["business_name"]
        .fillna("")
        .astype(str)
    )

    df["business_address"] = (
        df["business_address"]
        .fillna("")
        .astype(str)
    )

    df["country"] = (
        df["country"]
        .fillna("")
        .astype(str)
        .str.upper()
        .str.strip()
    )

    # -----------------------------------------------------
    # Extract structured fields BEFORE cleaning
    # -----------------------------------------------------

    df["postal_code"] = df["business_address"].apply(
        extract_postal_code
    )

    # -----------------------------------------------------
    # Clean text
    # -----------------------------------------------------

    df["clean_name"] = df["business_name"].apply(
        clean_str
    )

    df["clean_address"] = df["business_address"].apply(
        clean_str
    )

    # -----------------------------------------------------
    # Additional useful normalized fields
    # -----------------------------------------------------

    # Remove spaces from names for character-level matching
    df["name_compact"] = (
        df["clean_name"]
        .str.replace(" ", "", regex=False)
    )

    # Compact postal code
    df["postal_compact"] = (
        df["postal_code"]
        .str.replace(" ", "", regex=False)
        .str.replace("-", "", regex=False)
        .str.upper()
    )

    return df


# ---------------------------------------------------------
# Run preprocessing
# ---------------------------------------------------------

print("Preprocessing Source 1...")
df_s1_clean = preprocess_entities(df_s1)

print("Preprocessing Source 2...")
df_s2_clean = preprocess_entities(df_s2)

print("Preprocessing Source 3...")
df_s3_clean = preprocess_entities(df_s3)

print("\nPreprocessing Complete!")

print("\nSource 1:", df_s1_clean.shape)
print("Source 2:", df_s2_clean.shape)
print("Source 3:", df_s3_clean.shape)

print("\nExample:")
print(
    df_s1_clean[
        [
            "business_name",
            "clean_name",
            "business_address",
            "postal_code",
            "country"
        ]
    ].head()
)
```

```python
import numpy as np
from scipy.sparse import csr_matrix

def fast_tfidf_top_k(
    source_matrix,
    target_matrix,
    target_ids,
    top_k=20,
    batch_size=20000
):
    """
    Fast Top-K cosine candidate retrieval using sparse
    matrix multiplication.

    Assumes TF-IDF matrices are L2-normalized.
    """

    source_matrix = csr_matrix(source_matrix)
    target_matrix = csr_matrix(target_matrix)

    n_rows = source_matrix.shape[0]

    # Transpose once instead of repeatedly
    target_matrix_T = target_matrix.T.tocsc()

    all_candidate_ids = []

    for i in range(0, n_rows, batch_size):

        batch = source_matrix[i:i + batch_size]

        # Sparse cosine similarity
        sim_scores = batch.dot(target_matrix_T)

        for row_idx in range(sim_scores.shape[0]):

            row = sim_scores.getrow(row_idx).toarray().ravel()

            if len(row) > top_k:

                # Find Top-K without sorting the entire array
                top_indices = np.argpartition(
                    row,
                    -top_k
                )[-top_k:]

                # Sort only the Top-K
                top_indices = top_indices[
                    np.argsort(
                        -row[top_indices]
                    )
                ]

            else:
                top_indices = np.arange(len(row))

            all_candidate_ids.append(
                target_ids[top_indices]
            )

        processed = min(i + batch_size, n_rows)

        print(
            f"Processed {processed:,} / "
            f"{n_rows:,} rows "
            f"({100 * processed / n_rows:.1f}%)"
        )

    return all_candidate_ids
```

```python
import numpy as np
from scipy.sparse import csr_matrix

def fast_tfidf_top_k(
    source_matrix,
    target_matrix,
    target_ids,
    top_k=20,
    batch_size=20000
):
    """
    Fast Top-K cosine candidate retrieval using sparse
    matrix multiplication.

    Assumes TF-IDF matrices are L2-normalized.
    """

    source_matrix = csr_matrix(source_matrix)
    target_matrix = csr_matrix(target_matrix)

    n_rows = source_matrix.shape[0]

    # Transpose target matrix once
    target_matrix_T = target_matrix.T.tocsc()

    all_candidate_ids = []

    for i in range(0, n_rows, batch_size):

        batch = source_matrix[i:i + batch_size]

        # Sparse matrix multiplication
        sim_scores = batch.dot(target_matrix_T)

        for row_idx in range(sim_scores.shape[0]):

            # Convert only the current row to dense
            row = sim_scores.getrow(row_idx).toarray().ravel()

            if len(row) > top_k:

                # Get Top-K without sorting everything
                top_indices = np.argpartition(
                    row,
                    -top_k
                )[-top_k:]

                # Sort Top-K by similarity descending
                top_indices = top_indices[
                    np.argsort(
                        -row[top_indices]
                    )
                ]

            else:
                top_indices = np.arange(len(row))

            all_candidate_ids.append(
                target_ids[top_indices]
            )

        processed = min(i + batch_size, n_rows)

        print(
            f"Processed {processed:,} / "
            f"{n_rows:,} rows "
            f"({100 * processed / n_rows:.1f}%)"
        )

    return all_candidate_ids

print("fast_tfidf_top_k() is ready!")
```

```python
# ============================================================
# COMPLETE TF-IDF SETUP — RUN THIS ONE CELL
# ============================================================

import os
import re
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix
from sklearn.feature_extraction.text import TfidfVectorizer

# ------------------------------------------------------------
# 1. File paths
# ------------------------------------------------------------

S1_PATH = "/content/train_source1.tsv"
S2_PATH = "/content/train_source2.tsv"
S3_PATH = "/content/train_source3.tsv"

print("Checking files...")

for path in [S1_PATH, S2_PATH, S3_PATH]:
    print(path, "->", os.path.exists(path))

# ------------------------------------------------------------
# 2. Load datasets
# ------------------------------------------------------------

print("\nLoading datasets...")

df_s1 = pd.read_csv(
    S1_PATH,
    sep="\t",
    encoding_errors="replace",
    on_bad_lines="skip"
)

df_s2 = pd.read_csv(
    S2_PATH,
    sep="\t",
    encoding_errors="replace",
    on_bad_lines="skip"
)

df_s3 = pd.read_csv(
    S3_PATH,
    sep="\t",
    encoding_errors="replace",
    on_bad_lines="skip"
)

print("S1:", df_s1.shape)
print("S2:", df_s2.shape)
print("S3:", df_s3.shape)

# ------------------------------------------------------------
# 3. Cleaning function
# ------------------------------------------------------------

def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Keep letters, numbers and spaces
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # Remove multiple spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


def clean_source_dataframe(df):

    df = df.copy()

    # Business name
    df["business_name"] = (
        df["business_name"]
        .fillna("")
        .apply(clean_text)
    )

    # Address
    df["business_address"] = (
        df["business_address"]
        .fillna("")
        .apply(clean_text)
    )

    # Country
    df["country"] = (
        df["country"]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )

    # Postal code
    df["postal_code"] = (
        df["business_address"]
        .str.extract(r"\b(\d{4,10})\b", expand=False)
        .fillna("")
    )

    return df


# ------------------------------------------------------------
# 4. Clean all three sources
# ------------------------------------------------------------

print("\nCleaning datasets...")

df_s1_clean = clean_source_dataframe(df_s1)
df_s2_clean = clean_source_dataframe(df_s2)
df_s3_clean = clean_source_dataframe(df_s3)

print("Cleaning complete.")

print("\nS1 sample:")
print(df_s1_clean.head())

# ------------------------------------------------------------
# 5. Prepare business names
# ------------------------------------------------------------

s1_names = (
    df_s1_clean["business_name"]
    .fillna("")
    .astype(str)
)

s2_names = (
    df_s2_clean["business_name"]
    .fillna("")
    .astype(str)
)

s3_names = (
    df_s3_clean["business_name"]
    .fillna("")
    .astype(str)
)

# Combine names so the SAME TF-IDF vocabulary is used
all_names = pd.concat(
    [s1_names, s2_names, s3_names],
    ignore_index=True
)

print("\nTotal business names:", len(all_names))

# ------------------------------------------------------------
# 6. Build TF-IDF vectorizer
# ------------------------------------------------------------

print("\nBuilding character 3-gram TF-IDF vocabulary...")

vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3, 3),
    lowercase=False,
    dtype=np.float32
)

vectorizer.fit(all_names)

print("Vocabulary size:", len(vectorizer.vocabulary_))

# ------------------------------------------------------------
# 7. Transform each source
# ------------------------------------------------------------

print("\nCreating TF-IDF matrices...")

tfidf_s1 = vectorizer.transform(s1_names)
print("S1 TF-IDF created:", tfidf_s1.shape)

tfidf_s2 = vectorizer.transform(s2_names)
print("S2 TF-IDF created:", tfidf_s2.shape)

tfidf_s3 = vectorizer.transform(s3_names)
print("S3 TF-IDF created:", tfidf_s3.shape)

# ------------------------------------------------------------
# 8. Convert to CSR format
# ------------------------------------------------------------

tfidf_s1 = csr_matrix(tfidf_s1)
tfidf_s2 = csr_matrix(tfidf_s2)
tfidf_s3 = csr_matrix(tfidf_s3)

# ------------------------------------------------------------
# 9. Verify everything
# ------------------------------------------------------------

print("\n========================================")
print("TF-IDF SETUP COMPLETE")
print("========================================")

print("tfidf_s1:", tfidf_s1.shape)
print("tfidf_s2:", tfidf_s2.shape)
print("tfidf_s3:", tfidf_s3.shape)

print("\nNon-zero values:")
print("S1:", tfidf_s1.nnz)
print("S2:", tfidf_s2.nnz)
print("S3:", tfidf_s3.nnz)

print("\nSame vocabulary:")
print(
    tfidf_s1.shape[1]
    == tfidf_s2.shape[1]
    == tfidf_s3.shape[1]
)

print("\nVariables available:")
print("df_s1_clean:", "df_s1_clean" in globals())
print("df_s2_clean:", "df_s2_clean" in globals())
print("df_s3_clean:", "df_s3_clean" in globals())

print("tfidf_s1:", "tfidf_s1" in globals())
print("tfidf_s2:", "tfidf_s2" in globals())
print("tfidf_s3:", "tfidf_s3" in globals())
```

```python
# ============================================================
# FAST TF-IDF CANDIDATE GENERATION
# S1 -> S2 and S1 -> S3
# ============================================================

import numpy as np
from scipy.sparse import csr_matrix

# ------------------------------------------------------------
# Fast top-k retrieval function
# ------------------------------------------------------------

def fast_tfidf_top_k(
    source_matrix,
    target_matrix,
    target_ids,
    top_k=20,
    batch_size=5000
):

    source_matrix = csr_matrix(source_matrix)
    target_matrix = csr_matrix(target_matrix)

    n_rows = source_matrix.shape[0]
    top_k = min(top_k, target_matrix.shape[0])

    # Transpose target once
    target_matrix_T = target_matrix.T.tocsc()

    all_candidate_ids = []

    print(f"Source rows : {n_rows:,}")
    print(f"Target rows : {target_matrix.shape[0]:,}")
    print(f"Top-K       : {top_k}")
    print(f"Batch size  : {batch_size:,}")
    print()

    for start in range(0, n_rows, batch_size):

        end = min(start + batch_size, n_rows)

        batch = source_matrix[start:end]

        # Cosine similarity because TF-IDF is L2 normalized
        similarity_matrix = batch.dot(target_matrix_T)

        for row_idx in range(similarity_matrix.shape[0]):

            row = similarity_matrix.getrow(row_idx)

            # Only retrieve non-zero similarities
            indices = row.indices
            scores = row.data

            if len(indices) == 0:

                # No lexical overlap
                candidate_indices = np.array([], dtype=np.int64)

            elif len(indices) <= top_k:

                order = np.argsort(-scores)
                candidate_indices = indices[order]

            else:

                # Get top-k without sorting everything
                top_positions = np.argpartition(
                    scores,
                    -top_k
                )[-top_k:]

                # Sort those top-k by similarity
                top_positions = top_positions[
                    np.argsort(-scores[top_positions])
                ]

                candidate_indices = indices[top_positions]

            all_candidate_ids.append(
                target_ids[candidate_indices]
            )

        print(
            f"Processed {end:,} / {n_rows:,} "
            f"({100 * end / n_rows:.1f}%)"
        )

    return all_candidate_ids


# ============================================================
# S1 -> S2
# ============================================================

print("============================================")
print("GENERATING S1 -> S2 CANDIDATES")
print("============================================")

s2_candidates = fast_tfidf_top_k(
    tfidf_s1,
    tfidf_s2,
    df_s2_clean["entity_id"].values,
    top_k=20,
    batch_size=5000
)

print("\nS1 -> S2 candidate generation COMPLETE.")


# ============================================================
# Check S2 candidates
# ============================================================

print("\nS2 candidate information:")
print("Number of S1 records:", len(s2_candidates))
print("Candidates for first record:", len(s2_candidates[0]))
print("First S1 entity:", df_s1_clean.iloc[0]["entity_id"])
print("Candidate S2 IDs:", s2_candidates[0])
```

```python
# ============================================================
# GENERATE SENTENCE-TRANSFORMER EMBEDDINGS
# ============================================================

import os
import gc
import numpy as np

# ------------------------------------------------------------
# Make sure model exists
# ------------------------------------------------------------

from sentence_transformers import SentenceTransformer

if "model" not in globals():
    model = SentenceTransformer(
        "all-MiniLM-L6-v2",
        device="cpu"
    )

print("Model ready.")
print("Embedding dimension:", model.get_embedding_dimension())


# ------------------------------------------------------------
# Prepare text
# ------------------------------------------------------------

s1_texts = (
    df_s1_clean["business_name"]
    .fillna("")
    .astype(str)
    .tolist()
)

s2_texts = (
    df_s2_clean["business_name"]
    .fillna("")
    .astype(str)
    .tolist()
)

s3_texts = (
    df_s3_clean["business_name"]
    .fillna("")
    .astype(str)
    .tolist()
)


# ------------------------------------------------------------
# Embedding function
# ------------------------------------------------------------

def generate_embeddings(
    texts,
    batch_size=256,
    output_file=None
):

    print("\nGenerating embeddings...")
    print("Number of texts:", len(texts))
    print("Batch size:", batch_size)

    embeddings = []

    for start in range(0, len(texts), batch_size):

        end = min(start + batch_size, len(texts))

        batch_texts = texts[start:end]

        batch_embeddings = model.encode(
            batch_texts,
            batch_size=batch_size,
            show_progress_bar=False,
            convert_to_numpy=True,
            normalize_embeddings=True
        )

        embeddings.append(
            batch_embeddings.astype(np.float32)
        )

        if end % 10000 < batch_size or end == len(texts):
            print(
                f"Processed {end:,} / {len(texts):,} "
                f"({100 * end / len(texts):.1f}%)"
            )

    embeddings = np.vstack(embeddings)

    print("Final shape:", embeddings.shape)

    if output_file is not None:
        np.save(output_file, embeddings)

        print(
            "Saved:",
            output_file,
            f"({os.path.getsize(output_file) / (1024**2):.1f} MB)"
        )

    return embeddings


# ============================================================
# S1 EMBEDDINGS
# ============================================================

print("\n============================================")
print("S1 EMBEDDINGS")
print("============================================")

if os.path.exists("/content/s1_embeddings.npy"):

    print("Existing S1 embeddings found.")
    s1_embeddings = np.load("/content/s1_embeddings.npy")

    print("Loaded:", s1_embeddings.shape)

else:

    s1_embeddings = generate_embeddings(
        s1_texts,
        batch_size=256,
        output_file="/content/s1_embeddings.npy"
    )


# ============================================================
# S2 EMBEDDINGS
# ============================================================

print("\n============================================")
print("S2 EMBEDDINGS")
print("============================================")

if os.path.exists("/content/s2_embeddings.npy"):

    print("Existing S2 embeddings found.")
    s2_embeddings = np.load("/content/s2_embeddings.npy")

    print("Loaded:", s2_embeddings.shape)

else:

    s2_embeddings = generate_embeddings(
        s2_texts,
        batch_size=256,
        output_file="/content/s2_embeddings.npy"
    )


# ============================================================
# S3 EMBEDDINGS
# ============================================================

print("\n============================================")
print("S3 EMBEDDINGS")
print("============================================")

if os.path.exists("/content/s3_embeddings.npy"):

    print("Existing S3 embeddings found.")
    s3_embeddings = np.load("/content/s3_embeddings.npy")

    print("Loaded:", s3_embeddings.shape)

else:

    s3_embeddings = generate_embeddings(
        s3_texts,
        batch_size=256,
        output_file="/content/s3_embeddings.npy"
    )


# ============================================================
# FINAL CHECK
# ============================================================

print("\n============================================")
print("EMBEDDINGS COMPLETE")
print("============================================")

print("S1:", s1_embeddings.shape)
print("S2:", s2_embeddings.shape)
print("S3:", s3_embeddings.shape)

print("\nExpected:")
print("S1:", len(df_s1_clean))
print("S2:", len(df_s2_clean))
print("S3:", len(df_s3_clean))

print("\nDtypes:")
print("S1:", s1_embeddings.dtype)
print("S2:", s2_embeddings.dtype)
print("S3:", s3_embeddings.dtype)

print("\nAll embeddings normalized:",
      np.allclose(
          np.linalg.norm(s1_embeddings[:100], axis=1),
          1.0,
          atol=1e-4
      ))
```
